## Data Frames and Data Visualization

```python
import pandas as pd
import numpy as np
```

```python
data=pd.read_csv('/content/laptop_price - dataset.csv')
data.head(20)
```

```python
data.columns
```

```python
df=pd.DataFrame(data)
df
```

```python
data_subset= data[['Company', 'Product', 'TypeName', 'Inches', 'ScreenResolution', 'RAM (GB)', 'Memory', 'OpSys', 'Weight (kg)', 'Price (Euro)']]
data_subset
```

```python
data_subset.isna().sum()
```

```python
data_subset.describe()
```

```python
data_subset.dtypes
```

```python
data_subset.hist(bins=60, figsize=(20,10))
```

```python
pd.DataFrame(data_subset['RAM (GB)'].value_counts()).plot(kind='bar', figsize=(15,5))
```

```python
pd.DataFrame(data_subset['RAM (GB)'].value_counts(1)).plot(kind='bar', figsize=(15,5))
```

```python
pd.DataFrame(data_subset['Inches'].value_counts(1)).plot(kind='bar', figsize=(15,5))
```

```python
cost=data_subset[data_subset['Price (Euro)']==1000]
cost
```

```python
cost.plot(kind='scatter', x='Inches', y='Price (Euro)', figsize=(20,5))
```

```python
cost1=data_subset[data_subset['Weight (kg)']>1.0]
cost1.head(10)
```

```python
cost1.plot(kind='scatter', x='Product', y='Price (Euro)', figsize=(20,5))
```
